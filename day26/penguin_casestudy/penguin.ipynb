### average body mass of each species

```python
penguins[0]
```

```python
island_summary={}

for p in penguins:

    island=p["island"]

    body_mass_g=float(p["body_mass_g"])

    if island not in island_summary:

        island_summary[island]={"avg_body_mass":body_mass_g,"count":1}

    else:

        current_body_mass_g=island_summary[island]["avg_body_mass"]

        count=island_summary[island]["count"]

        island_summary[island]["avg_body_mass"]=(current_body_mass_g+body_mass_g)/count

        island_summary[island]["count"]+=1

island_summary
```
