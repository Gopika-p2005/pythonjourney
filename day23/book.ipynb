highest rating


```python
rating=[b["rating"] for b in books ]

max_rating=max(rating)

hight_rated_books=[b["title"] for b in books if b["rating"]==max_rating]

print(hight_rated_books)
```

lowest rating

```python
low_rating=min(rating)

lowest_rated_books=[b["title"] for b in books if b["rating"]==low_rating]

print(lowest_rated_books)
```
