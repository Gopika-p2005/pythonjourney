```python
movies = [
    {"title": "ABCD", "language": "Malayalam", "rating": 9, "year": 2011, "runtime": 120},
    {"title": "Vazha", "language": "Malayalam", "rating": 9, "year": 2022, "runtime": 140},
    {"title": "Kumbalangi Nights", "language": "Malayalam", "rating": 9, "year": 2019, "runtime": 135},
    {"title": "Super Deluxe", "language": "Tamil", "rating": 9, "year": 2019, "runtime": 176},
    {"title": "Jai Bhim", "language": "Tamil", "rating": 9, "year": 2021, "runtime": 164},
    {"title": "Parasite", "language": "Korean", "rating": 10, "year": 2019, "runtime": 132},
    {"title": "Spirited Away", "language": "Japanese", "rating": 10, "year": 2001, "runtime": 125},
    {"title": "The Godfather", "language": "English", "rating": 10, "year": 1972, "runtime": 175},
    {"title": "Inception", "language": "English", "rating": 9, "year": 2010, "runtime": 148},
    {"title": "Tumbbad", "language": "Hindi", "rating": 9, "year": 2018, "runtime": 104},
    {"title": "Dangal", "language": "Hindi", "rating": 9, "year": 2016, "runtime": 161},
    {"title": "Kantara", "language": "Kannada", "rating": 9, "year": 2022, "runtime": 148},
    {"title": "RRR", "language": "Telugu", "rating": 9, "year": 2022, "runtime": 187},
    {"title": "City of God", "language": "Portuguese", "rating": 10, "year": 2002, "runtime": 130},
    {"title": "Anatomy of a Fall", "language": "French", "rating": 9, "year": 2023, "runtime": 151}
]
```

```python
#all movie titles

all_movie_title=[m["title"] for m in movies ]

print(all_movie_title)
```

```python
#movie after 2005

after_2005=[m.get("title") for m in movies if m["year"]>2005] 

print(after_2005)
```

```python
#movies with rating 9

movie_rating_9=[m["title"] for m in movies if m["rating"]==9 ]

print(movie_rating_9)
```

```python
#movie runtime

all_movie_runtime={m["runtime"] for m in movies }

max_runtime=max(all_movie_runtime)

print(max_runtime)
```

```python
# hight runtime movie

hight_runtime_movie=[m["title"] for m in movies if m["runtime"]==max_runtime ]

print(hight_runtime_movie)
```

### Movies with lowest runtime



```python

lowest_runtime=min(all_movie_runtime)

lowest_runtime_movie=[m.get("title") for m in movies if m["runtime"]==lowest_runtime ]

lowest_runtime_movie
```

### Top rated movie


```python
movie_rating={m["rating"] for m in movies }

max_rating=max(movie_rating)

max_rating_movie=[m["title"] for m in movies if m["rating"]==max_rating]

max_rating_movie
```

### Lowest rating movie

```python
lowest_rating=min(movie_rating)

lowest_rating_movie=[m["title"] for m in movies if m["rating"]==lowest_rating ]

print(lowest_rating_movie)
```

malayalam movies

```python
malayalam_movies=[m["title"] for m in movies if m["language"]=="Malayalam"]

print(malayalam_movies)
```

year and movie count

```python
movie_year=[m["year"] for m in movies]

year_count={y:movie_year.count(y) for y in movie_year }

print(year_count)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,8))

x=list(year_count.keys())

y=list(year_count.values())

plt.bar(x,y)

plt.show()
```

languagewise movie count


```python
language=[m["language"] for m in movies]

language_wise_movie_count={l:language.count(l) for l in language }

print(language_wise_movie_count)
```
