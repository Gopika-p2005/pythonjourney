### finally

```python
n1=int(input("enter num1"))

n2=int(input("enter num2"))

try:

    print(n1/n2)

except Exception as e:

    n2=int(input("enter num2.."))

    try:

        print(n1/n2)

    except Exception as e:

        n2=int(input("enter num2"))

        print(n1/n2)

finally:

    print("data base commit...")

    print("file write....")
```

```python
try:

    password=input("enter the password..")

    if len(password)<6:

        raise Exception("password atleast 6 characters")
    

except Exception as e:

    print("error",e)

else:

    print("password accepted")

finally:

    print("passwordnvalidation completed")
```
