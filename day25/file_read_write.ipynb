### file read and write

```python
fr_items=open("C:\\Users\\Acer\\Desktop\\python works\\day25\\restaurant_menu.txt","r")

fw_veg=open("C:\\Users\\Acer\\Desktop\\python works\\day25\\veg.txt","w")

fw_nonveg=open("C:\\Users\\Acer\\Desktop\\python works\\day25\\nonveg.txt","w")

for line in fr_items:

    line=line.rstrip("\n")

    item,category=line.split(",")

    category=category.lstrip(" ")

    if category =="Veg":

        fw_veg.write(item+"\n")

    else:

        fw_nonveg.write(item+"\n")

fw_nonveg.close()

fw_veg.close()
```

```python
fr_movie=open("C:\\Users\\Acer\\Desktop\\python works\\day25\\movies.txt","r")

fw_comedy=open("C:\\Users\\Acer\\Desktop\\python works\\day25\\comedy.txt","w")

fw_action=open("C:\\Users\\Acer\\Desktop\\python works\\day25\\action.txt","w")

for line in fr_movie:

    line=line.rstrip("\n")
    
    title,categories=line.split(",")

    categories=categories.lstrip(" ")

    if categories=="Comedy":

        fw_comedy.write(title+"\n")

    else:

        fw_action.write(title+"\n")

fw_action.close()

fw_comedy.close()

```

```python
fr_langs_and_frames=open("C:\\Users\\Acer\\Desktop\\python works\\day25\\langs_and_frame.txt","r")

fw_framework=open("C:\\Users\\Acer\\Desktop\\python works\\day25\\framework.txt","w")

fw_library=open("C:\\Users\\Acer\\Desktop\\python works\\day25\\library.txt","w")

for line in fr_langs_and_frames:

    line=line.rstrip("\n")

    language,frame=line.split(",")

    frame=frame.lstrip(" ")

    if frame=="Framework":

        fw_framework.write(language+"\n")

    else:

        fw_library.write(language+"\n")

fw_library.close()

fw_framework.close()
```
