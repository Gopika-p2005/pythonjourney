### average spending in males

```python
sum=0

count=0

for c in customers:

    if c["gender"]=="Male":

        sum+=c["total_spent"]

        count+=1

        avg_male_spent=sum/count

print(avg_male_spent)
```

### average spending in female

```python
sum=0

count=0

for c in customers:

    if c["gender"]=="Female":

        sum+=c["total_spent"]

        count+=1

        avg_female_spent=sum/count

print(avg_female_spent)
```

### highest revenue city


```python
city_amount={}

for c in customers:

    city=c["city"]

    if city in city_amount:

        city_amount[city]+=c["total_spent"]

    else:

        city_amount[city]=c["total_spent"]


large=0

for k,v in city_amount.items():

    if v>large:

        large=v

        highest_city=k

print(highest_city,large)
```
