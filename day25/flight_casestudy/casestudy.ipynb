```python

fr=open("C:\\Users\\Acer\\Desktop\\python works\\day25\\flight_casestudy\\flights.cs.txt.xls","r")
```

### file convert into list of dictionary

```python
flights=[]

fr.readline()

for line in fr:

    line=line.rstrip("\n")

    year,month,passengers=line.split(",")

    dictionary={"year":year,"month":month,"passengers":passengers}

    flights.append(dictionary)

print(flights)
```

### year passengers dictionary

```python
result={}

for records in flights:

    year=records["year"]

    passengers_count=int(records["passengers"])

    if year in result:

        result[year]+=passengers_count

    else:

        result[year]=passengers_count

print(result)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=[12,8])


x=list(result.keys())

y=list(result.values())

plt.bar(x,y)

plt.xlabel("year")

plt.ylabel("passengers_count")

plt.show()
```

### month passengers dictionary

```python
results={}

for record in flights:

    month=record["month"]

    passenger_count=int(record["passengers"])

    if month in results:

        results[month]+=passenger_count

    else:

        results[month]=passenger_count

print(results)
```

```python

plt.figure(figsize=[12,8])

x=list(results.keys())

y=list(results.values())

plt.bar(x,y)

plt.xlabel("month")

plt.ylabel("passenger_count")

plt.show()
```
